```python
import numpy as np
import matplotlib.pyplot as plt

mat1 = np.fliplr(np.eye(4))
mat2 = np.zeros((4,4))
mat2[0,:]= 1
mat3 = np.zeros((4,4))
mat3[0,:]= 1
mat4 = np.eye(4)
h_s1 = np.hstack((mat1,mat2,mat3,mat4))

mat5= np.zeros((4,4))
mat5[:,0]=1
mat6 = np.zeros((4,4))
mat7 = np.zeros((4,4))
mat8= np.zeros((4,4))
mat8[:,3]= 1
h_s2 = np.hstack((mat5,mat6,mat7,mat8))

mat9= np.zeros((4,4))
mat9[:,0]=1
mat10 = np.zeros((4,4))
mat11 = np.zeros((4,4))
mat12= np.zeros((4,4))
mat12[:,3]= 1

h_s3 = np.hstack((mat9,mat10,mat11, mat12))

mat13 = np.eye(4)
mat14 = np.zeros((4,4))
mat14[3,:]= 1
mat15 = np.zeros((4,4))
mat15[3,:]= 1
mat16 = np.fliplr(np.eye(4))
h_s4 = np.hstack((mat13,mat14,mat15,mat16))

h_s= np.vstack((h_s1,h_s2,h_s3,h_s4))
h_s[4:6,4:6]= 1
h_s[4:6, 10:12]=1
h_s[11,5:11]= 1
h_s[10,4]=1
h_s[10,11]=1

from matplotlib.colors import LinearSegmentedColormap
black_yellow = LinearSegmentedColormap.from_list('black_yellow',['black', 'yellow'])
plt.imshow(h_s, cmap=black_yellow)
plt.axis('off')
plt.show()
print()
```
