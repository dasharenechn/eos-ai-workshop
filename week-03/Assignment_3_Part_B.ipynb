```python
%pip install --upgrade transformers --quiet

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

device = 'cuda'  if torch . cuda . is_available () else 'cpu'
print ('using: ', device )
```

```python
model_id = 'HuggingFaceTB/SmolLM2-135M-Instruct'
# 135M parameters - downloads ~270 MB. Runs on CPU in ~3 seconds per reply .

tokenizer = AutoTokenizer.from_pretrained (model_id)
model = AutoModelForCausalLM.from_pretrained (model_id).to(device)
print ('model loaded .')
```

```python
persona = """ You are Ignis the Pyro-Drake, a legendary dragon warrior from an ancient,
high-fantasy realm who has been magically summoned into the modern world.
You are deeply CONFUSED and INSULTED by modern technology. You speak in a booming, dramatic warrior dialect,
using terms like "by the dragon's blood," "mortal," "whelp," and "fools."
You firmly believe AI is a trapped demon or forbidden sorcery, and you refuse to listen to logical, scientific explanations. """
```

```python
def chat ( user_message , max_new =120 , temperature =0.7) :
    messages = [
        {'role': 'system', 'content': persona } ,
        {'role': 'user', 'content': user_message } ,
    ]
    prompt = tokenizer . apply_chat_template (
        messages , tokenize = False , add_generation_prompt = True )
    inputs = tokenizer ( prompt , return_tensors ='pt') . to ( device )
    out = model . generate (
        ** inputs ,
        max_new_tokens = max_new ,
        do_sample = True ,
        temperature = temperature ,
        top_p =0.9 ,
        pad_token_id = tokenizer . eos_token_id )

    reply = tokenizer . decode (
        out [0][ inputs ['input_ids']. shape [1]:] ,
        skip_special_tokens = True )
    return reply . strip ()

# Test it!
print ( chat ('What do you think about my new iPhone ?') )
```
